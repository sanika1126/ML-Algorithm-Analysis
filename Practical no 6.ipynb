```python
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
# Input: Study hours
X = np.array([1, 2, 3, 4, 5], dtype=float)

# Output: Marks
y = np.array([2, 4, 5, 4, 5], dtype=float)

print("Study Hours:", X)
print("Marks:", y)
```

```python
w = 0.0
b = 0.0

learning_rate = 0.01
epochs = 1000

n = len(X)

print("Initial weight:", w)
print("Initial bias:", b)
print("Learning rate:", learning_rate)
print("Epochs:", epochs)
```

```python
for i in range(epochs):
    y_pred = w * X
    error = y_pred - y

    gradient = (2 / n) * np.sum(error * X)

    w = w - learning_rate * gradient

print("Final Weight:", round(w, 2))
```

```python
prediction = w * X

print("Actual Values:", y)
print("Predicted Values:", np.round(prediction, 2))
```

```python
plt.scatter(X, y)
plt.plot(X, prediction)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Gradient Descent")

plt.show()
```

```python
prediction = w * X

print("Actual Values:", y)
print("Predicted Values:", np.round(prediction, 2))
```

```python
plt.scatter(X, y, label="Actual Values")
plt.plot(X, prediction, label="Predicted Values")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent Analysis")
plt.legend()
plt.grid()
plt.show()
```
